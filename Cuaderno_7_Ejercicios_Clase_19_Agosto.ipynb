```python
precios = np.array([24000, 28000, 22000, 30000, 8000])

precios[0]        # 24000        primer elemento
precios[-1]       # 8000         último elemento
precios[1:4]      # [28000 22000 30000]   el final se excluye

mascara = precios > 20000     # [True True True True False]
precios[mascara]              # [24000 28000 22000 30000]
precios[mascara].size         # 4

np.where(precios > 25000, "alto", "normal")
# ['normal' 'alto' 'normal' 'alto' 'normal']
```

```python
ingresos = np.array(
    [48000, 28000, 66000, 60000, 8000])

ingresos.sum()     # 210000
ingresos.mean()    # 42000.0
ingresos.max()     # 66000
ingresos.std()     # 21391.59
```
