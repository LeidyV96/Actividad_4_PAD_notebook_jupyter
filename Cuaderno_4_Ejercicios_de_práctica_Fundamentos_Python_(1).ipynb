# **Reto integrador**

**Clasificación de temperaturas**

Completa todos los espacios. Antes de ejecutar, prueba mentalmente tu solución con los valores: 15, 18, 22, 25 y 30.


```python
# El programa debe solicitar 5 temperaturas.
# Para cada una debe mostrar:
# "Baja" si t < 18
# "Normal" si 18 <= t <= 25
# "Alta" si t > 25
for t in range(5):
 t = float(input("Temperatura: "))
 if t<18:
    print("Baja")
 elif 18<=t<=5:
    print("Normal")
 else:
    print("Alta")
```
